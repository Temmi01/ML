5.

```python
new_data = data["CET"].str.split("-", expand=True)
new_data.columns = ["Year", "Month", "Day"]
new_data = new_data.astype(int)
data = data.drop(columns=["CET"])
data = pd.concat([new_data, data], axis=1)
data
```

6.

```python
not_null = data[data["Events"].isna()]
print("Кількість днів із порожнім значенням Events:", len(not_null))

min_value = data["Mean Humidity"].min()
min_rows = data[data["Mean Humidity"] == min_value]
print("День, у який середня вологість була мінімальною, а також швидкості вітру в цей день:")
display(min_rows[["Year", "Month", "Day", "Max Wind SpeedKm/h", "Mean Wind SpeedKm/h"]])

temp = data[(data["Mean TemperatureC"] >= 0) & (data["Mean TemperatureC"] <=5)]
months = temp["Month"].unique()
print("Місяці, коли середня температура від нуля до п’яти градусів:", months)
```

7.

```python
avg_temp = data.groupby(["Month", "Day"])["Max TemperatureC"].mean().reset_index()
avg_temp[["Max TemperatureC"]] = avg_temp[["Max TemperatureC"]].round(2)
print("Середня максимальна температура по кожному дню за всі роки:")
display(avg_temp)

fogs = data[data["Events"] == "Fog"]
fog_count = fogs.groupby("Year").size().reset_index(name="FogDays")
print("Кількість днів в кожному році з туманом: ")
fog_count
```

8.

```python
event_counts = data["Events"].value_counts()

plt.figure(figsize=(8,5))
event_counts.plot(kind="bar", color="blue")
plt.title("Кількість входжень кожного івенту")
plt.xlabel("Events")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

9.

```python
directions = ["N", "NE", "E", "SE", "S", "SW", "W", "NW"]
def choose_direction(deg):
  if deg > 337.5:
    return directions[0]
  
  for i in range(len(directions)):
    if deg - 22.5 - 45*(i) < 0:
      return directions[i]
 
data["WindDir"] = data["WindDirDegrees"].apply(choose_direction)
wind_counts = data["WindDir"].value_counts().reindex(directions, fill_value=0)

plt.figure(figsize=(6,6))
plt.pie(wind_counts, labels=wind_counts.index, startangle=143, counterclock=False)
plt.title("Розподіл напрямків вітру")
plt.show()
```

10.

```python
monthly_avg = data.groupby(["Year", "Month"]).agg({
    "Max TemperatureC": "mean",
    "Min DewpointC": "mean"
}).reset_index()

monthly_avg[["Max TemperatureC", "Min DewpointC"]] = monthly_avg[["Max TemperatureC", "Min DewpointC"]].round(2)

monthly_avg["YearMonth"] = monthly_avg["Year"].astype(str) + "-" + monthly_avg["Month"].astype(str)

plt.figure(figsize=(22,8))
plt.bar(monthly_avg["YearMonth"], monthly_avg["Max TemperatureC"], color="skyblue", width=0.9)
plt.xlabel("Рік-Місяць")
plt.ylabel("Max Temperature (°C)")
plt.title("Середня максимальна температура по місяцях кожного року")
plt.xticks([])
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()

# Графік 2: Середня Min Dewpoint
plt.figure(figsize=(22,8))
plt.bar(monthly_avg["YearMonth"], monthly_avg["Min DewpointC"], color="salmon", width=0.9)
plt.xlabel("Рік-Місяць")
plt.ylabel("Min Dewpoint (°C)")
plt.title("Середня мінімальна точка роси по місяцях кожного року")
plt.xticks([])
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```
